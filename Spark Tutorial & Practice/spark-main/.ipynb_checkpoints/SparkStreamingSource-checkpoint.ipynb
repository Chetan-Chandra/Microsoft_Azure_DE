```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
inputDirectory = "C:/Users/Admin/Documents/Udemy/Microsoft Azure Data Engineer/Spark Tutorial & Practice/spark-main/input/"
```

```python
# Create the SparkSession
spark = SparkSession\
        .builder\
        .appName("sparkstreamsource")\
        .getOrCreate()
```

```python
fileSchema = (StructType([\
                       StructField("userID", IntegerType(), True),
                       StructField("name", StringType(), True),
                       StructField("age",IntegerType(), True),
                       StructField("friends",IntegerType(), True),
                        ]))
```

```python
inputDF = (spark
          .readStream
          .format("csv")
          .schema(fileSchema)
          .load(inputDirectory))
```

```python
outputDir = "C:/Users/Admin/Documents/Udemy/Microsoft Azure Data Engineer/Spark Tutorial & Practice/spark-main/output/"
checkpointDir = "C:/Users/Admin/Documents/Udemy/Microsoft Azure Data Engineer/Spark Tutorial & Practice/spark-main/checkpoint/"
resultDF = inputDF
```

```python
streamingQuery = (resultDF.writeStream
.format("csv")
.option("path", outputDir)
.option("checkpointLocation", checkpointDir)
.start())
```
