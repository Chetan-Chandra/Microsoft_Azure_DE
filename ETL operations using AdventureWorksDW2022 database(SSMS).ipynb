## Step 3: Extract Data

```python
# Extract data from a table (e.g., DimProduct)
query = "SELECT * FROM DimProduct"
df_product = pd.read_sql(query, conn)

# Adjust pandas settings to display all columns
#pd.set_option('display.max_columns', None)  # None means no limit
#pd.set_option('display.width', None)        # No limit on width

# Check the extracted data
print(df_product.head())
```

## Step 4: Transform Data

```python
# Example transformation: Filter products with a specific condition
df_transformed = df_product[df_product['Color'] == 'Black']
```

```python
# Create a copy to avoid warnings
df_transformed = df_product[df_product['Color'] == 'Black'].copy()  
```

```python
df_transformed.head()
```

```python
# Additional transformation: Create a new column
df_transformed['Price'] = df_transformed['StandardCost'] * 1.2  # Example calculation
```

```python
# Use .loc to assign the new column
df_transformed.loc[:, 'Price'] = df_transformed['StandardCost'] * 1.2  
```

```python
df_transformed.head()
```

## Step 5: Load Data

#### Load Back to SQL Server

```python

# Load the transformed data back to a new table in the database
df_transformed.to_sql('TransformedProducts', conn, if_exists='replace', index=False)
```

#### Export to CSV

```python
# Export the transformed data to a CSV file
df_transformed.to_csv('TransformedProducts.csv', index=False)
```

## Full ETL Example

```python
import pandas as pd
import pyodbc

# Step 1: Connect to SQL Server
conn_string = (
    "DRIVER={SQL Server};"
    "SERVER=your_server_name;"
    "DATABASE=AdventureWorksDW2022;"
    "UID=your_username;"
    "PWD=your_password;"
)
conn = pyodbc.connect(conn_string)

# Step 2: Extract
query = "SELECT * FROM DimProduct"
df_product = pd.read_sql(query, conn)

# Step 3: Transform
df_transformed = df_product[df_product['Color'] == 'Black']
df_transformed['Price'] = df_transformed['StandardCost'] * 1.2

# Step 4: Load
# Load to SQL Server
df_transformed.to_sql('TransformedProducts', conn, if_exists='replace', index=False)

# Optionally, export to CSV
df_transformed.to_csv('TransformedProducts.csv', index=False)

# Close the connection
conn.close()
```
