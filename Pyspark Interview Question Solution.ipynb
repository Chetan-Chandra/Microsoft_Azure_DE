```python
# Step-by-Step Solution
# 1. Initialize Spark Session

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("CricketData") \
    .getOrCreate()

#Explanation: This initializes the Spark session, which is necessary for working with PySpark.
```

```python
# 2. Create DataFrames
# Players Dataset:

from pyspark.sql.types import StructType, StructField, StringType, IntegerType

schema = StructType([
    StructField("player", StringType(), True),
    StructField("runs", IntegerType(), True),
    StructField("50s/100s", StringType(), True)
])
```

```python

data = [
    ("Sachin-IND", 18694, "93/49"),
    ("Ricky-AUS", 11274, "66/31"),
    ("Lara-WI", 10222, "45/21"),
    ("Rahul-IND", 10355, "95/11"),
    ("Jhonty-SA", 7051, "43/5"),
    ("Hayden-AUS", 8722, "67/19")
]

players_df = spark.createDataFrame(data, schema)

#Explanation: Defines the schema and creates a DataFrame for the Players dataset.
```

```python
players_df.show()
```

```python
# Countries Dataset:

data1 = [("IND", "India"), ("AUS", "Australia"), ("WI", "WestIndies"), ("SA", "SouthAfrica")]

countries_df = spark.createDataFrame(data1, ["SRT", "country"])

#Explanation: Creates a DataFrame for the Countries dataset with country codes and names.
```

```python
countries_df.show()
```

```python
# You need to register these DataFrames as temporary views to use SQL queries.

players_df.createOrReplaceTempView("players")
countries_df.createOrReplaceTempView("countries")
```

```python
spark.sql("select * from players")
```

```python
select * from countries
```

```python
# Query to select all data from the "players" view
players_query = "SELECT * FROM players"
players_result = spark.sql(players_query)
```

```python
# 3. Transform the DataFrames
# Splitting Columns:

from pyspark.sql.functions import col, split

players_transformed_df = players_df.withColumn("player_name", split(col("player"), "-").getItem(0)) \
                                   .withColumn("SRT", split(col("player"), "-").getItem(1))

# Explanation: Uses split() to separate the player name from the country code. 
# split(col("player"), "-").getItem(0) extracts the player name
# split(col("player"), "-").getItem(1) extracts the country code.
```

```python
players_transformed_df.show()
```

```python
# Splitting 50s/100s:

players_transformed_df = (players_transformed_df.withColumn("runs_50s", split(col("50s/100s"), "/").getItem(0).cast("int")) \
                                                .withColumn("runs_100s", split(col("50s/100s"), "/").getItem(1).cast("int"))
                                                .withColumn("total_50s_100s", col("runs_50s") + col("runs_100s")))

# Explanation: Similarly, splits the 50s/100s column into two new columns (runs_50s and runs_100s) 
# and computes their sum in a new column total_50s_100s.
```

```python
players_transformed_df.show()
```

```python
# 4. Join the DataFrames

result_df = (
            players_transformed_df.join(countries_df, players_transformed_df["SRT"] == countries_df["SRT"]
            )
            .select(
                    col("player_name").alias("playername"),
                    col("country"),
                    col("runs"),
                    col("total_50s_100s").alias("sum"),
                    )
            .where(
                    col("total_50s_100s") > 90
                    )
            .orderBy(
                    "country")
                    )

# Explanation:
# Join: Performs an inner join between players_transformed_df and countries_df on the SRT field.
# Select: Chooses relevant columns and renames them for clarity.
# Filter: Keeps rows where the sum of 50s and 100s is greater than 90.
# Sort: Orders the results by country name.
```
